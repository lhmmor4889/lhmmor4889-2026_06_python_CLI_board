```python
print("== CLI 게시판 실행 ==")

# 게시글들을 저장할 리스트
articles = []
last_article_number = 0



while True: # 계속 하고 싶음. 명령어 계속 넣고 싶음.
    cmd = input("명령어 ) ").strip() # 공백 지우기 추가 양쪽

    if cmd == "exit":
        break

    elif cmd == "article write":
        last_article_number += 1
        title = input("제목 : ") # 담고
        content = input("내용 : ") # 담고

        # 게시글 정보를 딕셔너리로 만들어 리스트에 추가
        article = { # 다음단계 항상 순서를 기억하자... 순서대로 출력되는거니까. 인풋되고 나서 아티클에 담기는 거임. 내용들이. 조정하기 위한 부분이고 밑에 변수에 또 어펜드 해서 저장하는.control과 저장은 다르다.
            "id": last_article_number,
            "title": title,
            "content": content
        }
        articles.append(article) #articles.append(article)은 articles라는 이름의 리스트(목록) 맨 끝에 새로운 article 데이터를 추가하겠다. 아티클은 저장이 아닌 그냥 조립이고
        #이걸 저장을 하기 위해 아티클스 리스트를 추가해야만 한다. 1번째 하고나서 올라가면 1번째 자료들이 사라지니까.

        print("{}번 글이 생성 되었습니다.".format(last_article_number)) # 입력할때마다 숫자가 증가됨

    elif cmd == "article list":
        if not articles:
            print("등록된 게시글이 없습니다.")
            continue # 다시 하게끔 해야하니까. 넘어가게.

        print("========================================")
        print("번호    /    제목    /    내용")

        # 최신 글이 위로 오도록 리스트를 역순으로 순회
        for article in reversed(articles):  # 변수에 담고 아래 코드 실행! 즉, id 타이틀 콘탠트에 1번째 리스트 값을 넣어서 실행함.
            print("{} /    {} /    {}".format(article["id"], article["title"], article["content"])) # 딕셔너리 키 값을 넣어서 하면 벨류가 반환된다.
       #print(articles[0]), 이면 처음 추가되어 저장된 데이터가 호출된다.
       #print(articles[1]), 이면 두 번째 저장된 데이터가 출력된다.
      # reversed()가 원본을 유지한 채 새로운 역순 객체를 반환하는 것과 달리, .reverse()는 원본 리스트를 직접 수정(In-place)하며 반환값은 None입니다


        print("========================================")

    else:
        print("지원하지 않는 기능입니다.")

print("== CLI 게시판 종료 ==")
```

만약 articles라는 리스트에 세 개의 글이 담겨 있다면 다음과 같이 작동합니다.첫 번째 바퀴: 첫 번째 글을 꺼내 article 변수에 담고, 아래 코드를 실행합니다.두 번째 바퀴: 두 번째 글을 꺼내 article 변수에 담고, 아래 코드를 실행합니다.세 번째 바퀴: 세 번째 글을 꺼내 article 변수에 담고, 아래 코드를 실행합니다.



if not articles:는 "바구니가 비어 있다면(False라면)"이라는 조건문입니다.파이썬에서는 비어 있는 리스트 []를 False로 취급하기 때문에, not False가 되어 조건문 안의 코드가 실행됩니다.

```python
# article = [1, "123", "abc"]
print(article[0])  # 이게 번호인가?
print(article[1])  # 이게 제목인가? 내용인가?

# 전체 게시글을 담는 리스트
articles = []

# 게시글 작성 예시
# 순서대로 0번=번호, 1번=제목, 2번=내용으로 약속하고 넣어야 합니다.
article1 = [1, "123", "abc"]
articles.append(article1)

article2 = [2, "456", "def"]
articles.append(article2)

# 출력할 때
for article in reversed(articles):
    # 인덱스 번호(0, 1, 2)로만 값을 가져와야 해서 코드가 직관적이지 않음
    print("번호: {}, 제목: {}, 내용: {}".format(article[0], article[1], article[2]))
```

```python
# article = {"id": 1, "title": "123", "content": "abc"}
print(article["id"])       # 아, 번호구나!
print(article["title"])    # 아, 제목이구나!

# 전체 게시글을 담는 리스트
articles = []

# 게시글 작성 예시
# 순서 상관없이 Key(이름)를 지정해서 명확하게 넣습니다.
article1 = {
    "id": 1,
    "title": "123",
    "content": "abc"
}
articles.append(article1)

article2 = {
    "id": 2,
    "title": "456",
    "content": "def"
}
articles.append(article2)

# 출력할 때
for article in reversed(articles):
    # 인덱스 대신 "id", "title", "content"라는 이름으로 직접 꺼내 쓰므로 매우 직관적임
    print("번호: {}, 제목: {}, 내용: {}".format(article["id"], article["title"], article["content"]))
```
