```python
print("== CLI 게시판 실행 ==")

# 게시글들을 저장할 리스트
articles = []
last_article_number = 0

while True:
    cmd = input("명령어 ) ")

    if cmd == "exit":
        break

    elif cmd == "article write":
        last_article_number += 1
        title = input("제목 : ") # 담고
        content = input("내용 : ") # 담고

        # 게시글 정보를 딕셔너리로 만들어 리스트에 추가
        article = { # 다음단계 항상 순서를 기억하자... 순서대로 출력되는거니까. 인풋되고 나서 아티클에 담기는 거임. 내용들이.
            "id": last_article_number,
            "title": title,
            "content": content
        }
        articles.append(article) #articles.append(article)은 articles라는 이름의 리스트(목록) 맨 끝에 새로운 article 데이터를 추가하겠다

        print("{}번 글이 생성 되었습니다.".format(last_article_number))

    elif cmd == "article list":
        if not articles:
            print("등록된 게시글이 없습니다.")
            continue # 다시 하게끔 해야하니까. 넘어가게.

        print("========================================")
        print("번호    /    제목    /    내용")

        # 최신 글이 위로 오도록 리스트를 역순으로 순회
        for article in reversed(articles):
            print("{} /    {} /    {}".format(article["id"], article["title"], article["content"]))




        print("========================================")

    else:
        print("지원하지 않는 기능입니다.")

print("== CLI 게시판 종료 ==")
```
